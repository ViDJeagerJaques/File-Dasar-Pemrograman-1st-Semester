```python
i = 0
while i < 5:
  i += 1
  if i == 4:
    continue
  print(f"saya sudah {i} kali makan")

print("namun yang ke 4 saya masih belum kenyang")
```

```python
aktif = True # Variabel kontrol

while aktif:
  print("\n --- Menu Aplikasi --- ")
  print("1. Tambah Data")
  print("2. Lihat Laporan")
  print("3. Keluar")

pilihan = input("Masukkan pilihan (1-3): ")

if pilihan == '3':
   # Mekanisme mengubah kondisi menjadi False untuk menghentikan Loop
  aktif = False
  print("Terima kasih, program diakhiri.")
elif pilihan == '1':
   print("Anda memilih Tambah Data ... ")
elif pilihan == '2':
   print("Anda memilih Lihat Laporan ... ")
else:
   print("Pilihan tidak valid.")
```
