```python
# Iterasi menggunakan range(len(list))
daftar_nilai = [80,75,90]
total = 0
for indeks in range(len(daftar_nilai)):
# Mengakses item berdasarkan indeks
  total = total + daftar_nilai[indeks]
print("Total Nilai:", total)
```

```python
data_ujian = [85, 70, 92, 65, 88]
jumlah_data = len(data_ujian)
total_nilai = 0
# Iterasi dan Akumulasi
for nilai in data_ujian:
  total_nilai = total_nilai + nilai
# Perhitungan Akhir
rata_rata = total_nilai / jumlah_data
print(f"Total data: {jumlah_data}")
print(f"Total Nilai Akumulasi: {total_nilai}")
print(f"Rata-rata Kelas: {rata_rata :.2f}")
```

```python
nilai = int(input("Masukkan Nilai Akhir: "))
if nilai >= 80:
  grade ="A"
elif nilai >= 70:
  grade = "B"
elif nilai >= 60:
  grade = "C"
elif nilai >= 50:
  grade = "D"
else:
# Ini adalah kondisi default jika semua di atas False
  grade = "E"
print(f"Nilai Anda {nilai}, Grade Anda: {grade}")
```

```python
target = 42
data = [10, 20, 42, 50, 60]
print("Mulai proses pencarian ... ")
for angka in data:
  if angka == target:
    print(f"Data {target} DITEMUKAN!")
    break # Langsung keluar dari loop
print(f"Sedang memproses angka: {angka}")
# Program berlanjut di sini
print("Proses pencarian selesai.")
```

```python
aktif = True # Variabel kontrol
while aktif:
  print("\n --- Menu Aplikasi --- ")
  print("1. Tambah Data")
  print("2. Lihat Laporan")
  print("3. Keluar")
  pilihan = input("Masukkan pilihan (1-3): ")
  if pilihan == '3':
# Mekanisme mengubah kondisi menjadi False untuk menghentikan Loop
    aktif = False
    print("Terima kasih, program diakhiri.")
  elif pilihan == '1' :
    print("Anda memilih Tambah Data ... ")
  elif pilihan == '2':
    print("Anda memilih Lihat Laporan ... ")
  else:
    print("Pilihan tidak valid.")
```

```python
daftar_angka = [10, -5, 20, -1, 30]
total_positif = 0
for angka in daftar_angka:
  if angka < 0:
    print(f"Melewati angka negatif: {angka}")
    continue # Lewati baris kode di bawah, lanjut ke angka berikutnya
# Blok ini hanya dijalankan jika angka >= 0
total_positif = total_positif + angka
print("Total angka positif:", total_positif) # Output: 6
```
