```python
import re
teks = "Hubungi saya di 0812-3456-790"
# Pola untuk mencari angka: \d
pola = r"\d{4}-\d{4}-\d{4}"
hasil = re.findall(pola, teks)
print(hasil) # Output: ['0812-3456-7890']
```

```python
kontak = "admin@kampus.ac.id"
# Cara 1: String Methods
username = kontak.split("@") [0]
domain = kontak.split("@")[1]
# Cara 2: Regex (Untuk validasi simpel)
import re
if re.search(r"@[\w.]+", kontak):
  print("Format email valid")

print(f"User: {username}, Domain: {domain}")
```
