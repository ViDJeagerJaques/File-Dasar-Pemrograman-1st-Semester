```python
def hitung_luas_persegi (sisi):
  luas = sisi * sisi
  return luas # Mengembalikan nilai luas
#Baris kode setelah return tidak akan dieks
# Menerima nilai balik
hasil_luas = hitung_luas_persegi (5)
print(hasil_luas) # Output: 25
```

```python
def hitung_bmi (berat_kg, tinggi_m):
  """Fungsi menghitung BMI dan mengembalikan nilainya."""
  #Pastikan tinggi tidak nol untuk menghindari error
  if tinggi_m == 0:
    return 0
# Proses perhitungan
  bmi = berat_kg / (tinggi_m ** 2)
  return bmi
# Pemanggilan dan Penggunaan Nilai Balik
berat_saya = 70
tinggi_saya = 1.70
#Nilai 70 dan 1.70 adalah ARGUMEN
hasil = hitung_bmi (berat_saya, tinggi_saya)
print(f"BMI Anda adalah: {hasil:.2f}")
```

```python
berat_saya = float(input("Masukkan Berat Anda : "))
tinggi_saya = float(input("Masukkan Tinggi Anda (contoh 1.67): "))



def hitung_bmi (berat_kg, tinggi):

  if tinggi == 0:
    return 0

  bmi = berat_kg / (tinggi ** 2)
  return bmi

hasil = hitung_bmi (berat_saya, tinggi_saya)
print(f"BMI Anda adalah: {hasil:.2f}")

if hasil > 30:
  print("Obesitas")
elif hasil >= 23:
  print("Berat badan berlebih")
elif hasil <= 18.5:
  print("Berat badan kurang")
else:
  print("Berat badan ideal")
```

```python
counter = 10
def tambah_global():
  global counter # Mengacu pada variabel 'counter' yang global counter counter 1
  counter = counter + 1
  print("Counter di dalam:", counter)
tambah_global() # Output: 1
tambah_global() # Output: 2
```

```python
def hitung_volume_balok(panjang, lebar, tinggi):
  volume = panjang * lebar * tinggi
  return volume

hitung_luas_lol = hitung_volume_balok(8, 9, 11)
print(hitung_luas_lol)
```
