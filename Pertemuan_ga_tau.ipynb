```python
import datetime
def catat_aktivitas (pesan):
  waktu = datetime.datetime.now()
# Gunakan mode 'a' agar data ditambahkan di akhir (tidak menimpa)
  with open("log.txt", "a") as file_log:
    file_log.write(f"[{waktu}] - {pesan}\n")

catat_aktivitas ("User Login")
catat_aktivitas ("User Membuka Menu Laporan")

print("Aktivitas berhasil dicatat ke log.txt")
```
