```python
nilai = int(input("Masukkan nilai anda: "))

if nilai >= 80:
  grade = "A"
elif nilai >= 70:
  grade = "B"
elif nilai >= 60:
  grade = "C"
elif nilai >= 50:
  grade = "D"
else:
  grade = "E"

print(f"Nilai anda:{nilai}, Grade anda: {grade}")
```

```python
username = input("Masukkan username: ")
password = input("Masukkan password: ")

if username == "abc":
  if password == "123":
    print("Login berhasil")

    role = input("Masukkan role(Admin/User): ")
    if role == "Admin" or role == "admin" or role == "a" or role == "A":
      print("Akses penuh : Selamat datang di akses penuh Admin")

    else:
      print(f"Akses standar : Selamat datang username {username}")
  else:
    print("Password salah")
else:
  print("Username tidak ditemukan")

```
