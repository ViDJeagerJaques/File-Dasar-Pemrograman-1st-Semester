```python
stok_barang = {
  "Laptop": 10,
  "Mouse": 50,
  "Keyboard":30
}
harga_barang = {
  "Laptop": 15000000,
  "Mouse": 150000,
  "Keyboard": 300000
}

total_aset = 0
for barang, jumlah in stok_barang.items():
  nilai = jumlah*harga_barang [barang]
  total_aset += nilai
  print(f"Nilai stok {barang}: Rp{nilai:,}")
  print(f"\nTOTAL ASET TOKO: Rp{total_aset:,}")
```

```python
# MAINAN PERTAMA: ANJING 🐕

class Anjing:
    """Ini cetakan untuk bikin anjing"""

    def __init__(self, nama, warna):
        """Ini seperti memberi nama dan warna ke anjing"""
        self.nama = nama
        self.warna = warna

    def gonggong(self):
        """Anjing bisa gonggong"""
        print(f"{self.nama} bilang: Guk guk guk!")

    def main(self):
        """Anjing bisa main"""
        print(f"{self.nama} si anjing {self.warna} lagi main bola!")


# Bikin anjing pertama
anjing1 = Anjing("Brownie", "coklat")
anjing1.gonggong()
anjing1.main()

print()  # Baris kosong

# Bikin anjing kedua
anjing2 = Anjing("Snowy", "putih")
anjing2.gonggong()
anjing2.main()

print("\n" + "="*50 + "\n")

# MAINAN KEDUA: MOBIL 🚗

class Mobil:
    """Ini cetakan untuk bikin mobil"""

    def __init__(self, warna, merek):
        """Mobil punya warna dan merek"""
        self.warna = warna
        self.merek = merek
        self.kecepatan = 0  # Awalnya mobil diam

    def jalan(self):
        """Mobil jalan makin cepat"""
        self.kecepatan += 10
        print(f"Mobil {self.merek} {self.warna} jalan... Kecepatan: {self.kecepatan} km/jam")

    def berhenti(self):
        """Mobil berhenti"""
        self.kecepatan = 0
        print(f"Mobil {self.merek} berhenti!")

    def bunyi_klakson(self):
        """Mobil bunyikan klakson"""
        print(f"Mobil {self.merek}: Tin tin tin!")


# Bikin mobil merah
mobil_merah = Mobil("merah", "Ferrari")
mobil_merah.jalan()
mobil_merah.jalan()
mobil_merah.bunyi_klakson()
mobil_merah.berhenti()

print()

# Bikin mobil biru
mobil_biru = Mobil("biru", "Toyota")
mobil_biru.jalan()
mobil_biru.bunyi_klakson()

print("\n" + "="*50 + "\n")

# MAINAN KETIGA: KOTAK HARTA KARUN 🎁

class KotakHarta:
    """Kotak untuk nyimpen mainan"""

    def __init__(self, pemilik):
        """Kotak punya pemilik"""
        self.pemilik = pemilik
        self.isi = []  # Kotak masih kosong

    def masukkan(self, mainan):
        """Masukin mainan ke kotak"""
        self.isi.append(mainan)
        print(f"{self.pemilik} masukin {mainan} ke dalam kotak")

    def lihat_isi(self):
        """Lihat apa aja yang ada di kotak"""
        print(f"\nKotak {self.pemilik} isinya:")
        if len(self.isi) == 0:
            print("  Kosong nih!")
        else:
            for mainan in self.isi:
                print(f"  - {mainan}")


# Bikin kotak untuk Budi
kotak_budi = KotakHarta("Budi")
kotak_budi.masukkan("mobil-mobilan")
kotak_budi.masukkan("boneka")
kotak_budi.masukkan("bola")
kotak_budi.lihat_isi()

print()

# Bikin kotak untuk Siti
kotak_siti = KotakHarta("Siti")
kotak_siti.masukkan("boneka")
kotak_siti.masukkan("pensil warna")
kotak_siti.lihat_isi()
```
