```python
#d.	Praktekan dan jelaskan soal berikut:
data_mhs = [
    ['Abdi', 'Laki-laki', 172, 60, "Jakarta"],
    ['Budi', 'Laki-laki', 165, 60, "Bandung"],
    ['Caca', 'Perempuan', 150, 45, "Bali"],
    ["Supri", "Laki-laki", 167, 72, "Pekanbaru"]
]
#Tampilkan semua data yang ada menggunakan iterasi
for baris in data_mhs:
  nama = baris[0]
  kelamin = baris[1]
  Tinggi = baris[2]
  Berat = baris[3]
  Alamat = baris[4]
  print(f"Nama: {nama}\nJenis kelamin: {kelamin}\nTinggi:{Tinggi}\nBerat:{Berat}\nAlamat:{Alamat}\n")
```
